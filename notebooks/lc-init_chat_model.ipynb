```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

model = init_chat_model("deepseek-chat")
print(type(model))
```

```python
# 自定义模型提供商通过 兼容 OpenAI 格式的 Chat API 调用模型 demo

aliyun_base_url = os.getenv("ALIYUN_BASE_URL")
aliyun_api_key = os.getenv("ALIYUN_API_KEY")

print("aliyun_base_url::", aliyun_base_url)
print("aliyun_api_key::", aliyun_api_key)

bailian_chat_model = init_chat_model(
    model="qwen3.7-flash-2026-07-15",
    model_provider="openai",
    base_url=aliyun_base_url,
    api_key=aliyun_api_key,
)
print(type(model))

res = bailian_chat_model.invoke("你是什么模型？")
res1 = bailian_chat_model.invoke([{"role": "user", "content": "你是什么模型？"}])

print(res.content)
print(res1.content)

```

```python
streamCont = bailian_chat_model.stream("社会主义核心价值观有哪些？")
# 打印流式分片文本 chunk.content，打印完不换行，并且强制立刻输出到控制台，实现像聊天框那样一字一字实时输出的效果。
for chunk in streamCont:
    print(chunk.content, end="", flush=True)
```

将 chat_model 和 agent 绑定

```python

from langchain.agents import create_agent

ali_agent = create_agent(
    model=bailian_chat_model,
    system_prompt="You are an AI assistant that helps people find information.",
)

result = ali_agent.invoke({"messages": [{"role": "user", "content": "杭州有啥好玩的地方推荐？"}]})
# 打印结果结果
print("result::: ", result["messages"][-1].content)
```

```python
# 流式访问 messages: 最常用聊天打字机效果
messages = ali_agent.stream({"messages": [{"role": "user", "content": "杭州有啥好玩的地方推荐？"}]}, stream_mode="messages")
# 循环输出打字机效果
for chunk in messages:
    # stream_mode=messages 返回元组 (AIMessageChunk, metadata)
    msg_chunk, meta = chunk
    if msg_chunk.content:
        print(msg_chunk.content, end="", flush=True)
```
