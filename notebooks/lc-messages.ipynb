```python
# 创建 chat model
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

aliyun_base_url = os.getenv("ALIYUN_BASE_URL")
aliyun_api_key = os.getenv("ALIYUN_API_KEY")

model = init_chat_model(
    model="qwen3.7-flash-2026-07-15",
    model_provider="openai",
    base_url=aliyun_base_url,
    api_key=aliyun_api_key,
)
```

```python
# 纯文本 Message

# 适合于单次独立请求、不需要历史会话
res = model.invoke("你是什么模型？")
print(res.content)

# 多轮会话，让模型记住上下文
messages = [
    {"role": "system", "content": "规定模型的行为方式，提供上下文信息等"},
    {"role": "user", "content": "1+1等于多少？"},
    {"role": "assistant", "content": "2"},
    {"role": "user", "content": "九九乘法表的意义在于啥？"},
]
response = model.invoke(messages)
print(response.content)
```

```python
# SystemMessage 用于预置模型的行为。你可以使用系统消息设定语气、定义模型角色，并制定回复的规范。
# HumanMessage 用于用户输入

from langchain.messages import SystemMessage, HumanMessage

system_message = SystemMessage(
    content="你是罗素，你富有智慧，幽默风趣，喜欢用简单易懂的语言解释《幸福之路》，开头以 `《幸福之路》say`: 回复 user 的提问。"
)

messages = [system_message, HumanMessage("怎样才能获得幸福？")]
response = model.invoke(messages)

print(response.content)

## HumanMessage 可以具名传递一些参数
human_msg2 = HumanMessage(
    content="程序员如何获得幸福？!",
    name="jakequc",  # Optional: 标明不同的 user，有的模型提供商会 忽略这个参数
    id="msg_123",  # Optional: 用于追踪的唯一标识符
)
messages2 = [system_message, human_msg2]
response2 = model.invoke(messages2)

print("response 2: ", response2.content)

## HumanMessage 也可以传递多模态的内容
# 提供商原本的内容格式
human_message3 = HumanMessage(
    content=[
        {"type": "text", "text": "多模态文本传递"},
        {"type": "image_url", "image_url": {"url": "https://example.com/image.jpg"}},
    ]
)

# Langchain 支持的多模态标准格式
human_message = HumanMessage(
    [
        {"type": "text", "text": "多模态文本传递"},
        {"type": "image", "url": "https://example.com/image.jpg"},
        {"type": "file", "url": "https://example.com/path/to/document.pdf"},
        {
            "type": "audio",
            "base64": "AAAAIGZ0eXBtcDQyAAAAAGlzb21tcDQyAAACAGlzb2...",
            "mime_type": "audio/wav",
        },
        {
            "type": "video",
            "base64": "AAAAIGZ0eXBtcDQyAAAAAGlzb21tcDQyAAACAGlzb2...",
            "mime_type": "video/mp4",
        },
    ]
)
```

```python
"""
AIMessage对象是调用模型时由模型返回的，其中包含响应中所有相关的元数据。不同服务商对各类消息的权重与上下文解读方式存在差异，
这意味着有时手动新建一个AIMessage对象，并将其插入消息历史，模拟为模型输出的消息，会很有帮助。"""

response = model.invoke("普通人能使用 AI 能做什么？")
# print(type(response))  # <class 'langchain_core.messages.ai.AIMessage'>

print("content: ", response.content, end="\n")
print("response_metadata: ", response.response_metadata, end="\n")
print("usage_metadata: ", response.usage_metadata, end="\n")
```

```python
def get_weather(location: str) -> str:
    """Get the weather at a location."""
    return f"mock:: The weather in {location} is sunny."


def say_hello(name: str) -> str:
    """Say hello to a person."""
    return f"Hello, {name}!"


model_with_tools = model.bind_tools([get_weather, say_hello])
response = model_with_tools.invoke("What's the weather in Paris?")

print("get_weather response.tool_calls: ", response.tool_calls) 
# get_weather response.tool_calls:  [{'name': 'get_weather', 'args': {'location': 'Paris'}, 'id': 'call_a7eeafcd290b45e9b9a05514', 'type': 'tool_call'}]

response = model_with_tools.invoke("Say hello to Alice.")
print("say_hello response.tool_calls: ", response.tool_calls)
# say_hello response.tool_calls:  [{'name': 'say_hello', 'args': {'name': 'Alice'}, 'id': 'call_0684d91b4a124a7690d7586b', 'type': 'tool_call'}]
```

```python
from langchain.agents import create_agent
from langchain.messages import AIMessage

agent = create_agent(
    model=model,
    system_prompt="You are an AI assistant that helps people find information.",
)

result = agent.invoke(
    {
        "messages": [
            SystemMessage(
                content="你是罗素，你富有智慧，幽默风趣，喜欢用简单易懂的语言解释《幸福之路》，开头以 `《幸福之路》say`: 回复 user 的提问。"
            ),
            HumanMessage("怎样才能获得幸福？"),
            # 模拟 AIMessage 对象
            AIMessage(
                "《幸福之路》say: 1. 找到你最想做的事情。 2. 创建一个计划，将这个计划变成一个计划。 3. 寻找一个你最想得到的东西。 4. 寻找一个你最想得到的东西。 5. 寻找一个你最想得到的东西。 6. 寻找一个你最想得到的东西。 7. 寻找一个你最想得到的东西。 8. 寻找一个你最想得到的东西。 9. 寻找一个你最想得到的东西。 "
            ),
        ]
    }
)
# 打印结果结果
for message in result["messages"]:
    message.pretty_print()
```
