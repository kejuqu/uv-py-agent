# 本篇章主要是创建一个 agent

```python
# 1. 统一导入需要用到的包，加载环境变量： 使用 .env 和 dotenv 的形式
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
from langgraph.checkpoint.memory import InMemorySaver

load_dotenv()
```

```python
# 2. 定义工具

@tool
def get_current_weather(location: str) -> str:
    """Get the current weather in a given location"""
    return f"Current weather in {location} is 70 degrees and sunny"
```

```python
# 3. 创建 Agent 并注册 tool

# 3.1 定义系统提示词 system_prompt
system_prompt = """
 You are a wheather agent, You can only use the tool to get weather information.
 如果输入的指令无法完成，请返回 "I don't know"\n
"""

# 3.2 创建一个内存保存器，给你的智能体添加记忆，以在多次交互之间保持状态。这能让智能体记住过往对话与上下文。
checkpointer = InMemorySaver()

agent = create_agent(
    model="deepseek-flash",
    tools=[get_current_weather],
    system_prompt=system_prompt,
    checkpointer=checkpointer,
)
```

```python
# 发起调用
print("🚀 Starting agent...")
result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)
print(result["messages"][-1].content_blocks)
```
